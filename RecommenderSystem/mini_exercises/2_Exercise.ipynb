Step 1: Encode the genres using TF-IDF vectorization.

About TF-IDF: Term Frequency-Inverse Document Frequency

* TF (Term Frequency): How often a word appears in a document
* IDF (Inverse Document Frequency): How rare a word is across all documents (higher is rarer)
* TF-IDF = TF x IDF
* High score highlights words that are frequent in a document but rare across the corpus

Scores the importance of a word in a document (e.g., item description) relative to a collection (corpus)

It is a way of encoding textual information (in this case genres)
By encoding a movie using TF-IDF, a defining genre (e.g., only this movie has it) will get a higher score
